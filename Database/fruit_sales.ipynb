```python
import sqlite3
import pandas as pd
```

```python
# Connect (creates file if not exists)
conn = sqlite3.connect("fruit_sales.db")
cursor = conn.cursor()
```

```python
# Drop table if it exists
cursor.execute("DROP TABLE IF EXISTS fruit_sales")
```

```python
# Create table
cursor.execute("""
CREATE TABLE IF NOT EXISTS fruit_sales (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    product TEXT,
    quantity INTEGER,
    price REAL
)
""")
```

```python
df = pd.read_sql_query("SELECT * FROM fruit_sales", conn)
df
```
